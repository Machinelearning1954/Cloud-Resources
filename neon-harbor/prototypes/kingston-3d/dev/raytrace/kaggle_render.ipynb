```python
import glob, os
src = glob.glob('/kaggle/input/**/scene-*.json', recursive=True)
print(src)
assert src, 'add the dataset with the scene files first'
```

```python
# Upload dev/raytrace/render.py to /kaggle/working first, then:
import subprocess
for s in src:
    name = os.path.basename(s)[:-5]
    subprocess.run(['python3', '/kaggle/working/render.py', s, f'/kaggle/working/{name}', '--device', 'gpu', '--samples', '256', '--size', '1080x1920', '--views', 'hero,quay,aerial,street'], check=True)
```

```python
import shutil
shutil.make_archive('/kaggle/working/renders', 'zip', '/kaggle/working')
print('done: download /kaggle/working/renders.zip from the Output panel')
```
